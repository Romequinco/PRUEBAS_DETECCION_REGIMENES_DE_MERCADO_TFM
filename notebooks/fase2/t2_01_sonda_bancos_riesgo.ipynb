Objetivo: seleccionar comunicaciones oficiales que ayuden a interpretar el régimen macrofinanciero y los cambios de expectativas en **Europa, América y Asia**. El universo de trabajo son bolsas, deuda y divisas de mercados desarrollados y grandes emergentes. Se priorizan la política monetaria, las perspectivas de inflación y crecimiento, la liquidez y el riesgo sistémico.

Una institución grande no convierte todas sus publicaciones en información útil para invertir. Se conservan decisiones, comparecencias, actas, proyecciones y unos pocos informes de referencia. La selección distingue **comunicación monetaria**, con capacidad de cambiar expectativas, y **contexto de riesgo**, de menor frecuencia. Ninguna publicación tiene impacto garantizado: importa la sorpresa respecto a lo que ya descontaban los precios.

Este notebook prueba adquisición, fechas y extracción. **No identifica por sí solo el tema dominante del mercado ni demuestra capacidad predictiva**. Para ello habrá que contrastar los textos con expectativas y reacciones de precios usando únicamente información disponible en cada momento.

## Configuración y selección efectiva

Ejecuta la configuración y después los apartados deseados. Se utilizan los adaptadores existentes, pero la lista permitida se define **dentro de este notebook**: únicamente esas familias llegan a `sonda.ejecutar`. El inventario compartido y los demás notebooks no cambian.

`MODO_RAPIDO=True` limita la muestra a tres documentos por familia y cuarenta páginas por PDF. `USAR_CACHE=False` solicita contenido de nuevo respetando robots.txt y los controles del cliente. Para repetir una muestra sin refrescar sus originales, usa `True`. El corte se fija en UTC al ejecutar; puede reemplazarse por una fecha histórica explícita.

Las salidas de la revisión anterior se han retirado: incluían fuentes que ya no pertenecen a esta selección. Las nuevas mediciones aparecen al ejecutar. Las exportaciones usan el subdirectorio `mercados_relevantes` del corpus, para no sobrescribir las mediciones de la sonda general.

```python
from copy import deepcopy
from pathlib import Path
import sys

import pandas as pd
from IPython.display import display

RAIZ = next((p for p in [Path.cwd(), *Path.cwd().parents]
             if (p / "src/ingesta/bancos_riesgo/config/sonda_fuentes.json").exists()), None)
if RAIZ is None:
    raise RuntimeError("Abre el notebook desde este repositorio.")
if str(RAIZ / "src") not in sys.path:
    sys.path.insert(0, str(RAIZ / "src"))

from ingesta.bancos_riesgo.sonda import ConfigSonda, Sonda
from ingesta.bancos_riesgo.sonda.analysis import mostrar_apartado
from ingesta.comun.rutas import corpus_dir

pd.set_option("display.max_columns", 24)
pd.set_option("display.max_colwidth", 100)

MODO_RAPIDO = True
USAR_CACHE = False
CORTE_UTC = pd.Timestamp.now(tz="UTC").isoformat()

cfg = ConfigSonda(
    raiz=RAIZ,
    max_por_familia=3 if MODO_RAPIDO else 10,
    max_filas=3 if MODO_RAPIDO else 10,
    max_paginas=40 if MODO_RAPIDO else 200,
    usar_cache=USAR_CACHE,
    corte_utc=CORTE_UTC,
    directorio_corpus=corpus_dir("bancos_riesgo") / "mercados_relevantes",
)
sonda = Sonda(cfg)
```

```python
# Claves explícitas del inventario compartido: origen, institución, publicación.
# No se usan índices de fila ni filtros que incorporen familias nuevas sin revisión.
SELECCION = {
    1: [
        (1, "Fed", "Comunicado y decisión FOMC"),
        (1, "Fed", "Actas FOMC"),
        (1, "Fed", "Rueda de prensa: transcripción"),
        (1, "Fed", "Beige Book"),
        (1, "Fed", "Discurso"),
    ],
    2: [
        (2, "BCE", "Decisión de política monetaria"),
        (2, "BCE", "Declaración de política monetaria"),
        (2, "BCE", "Preguntas y respuestas"),
        (2, "BCE", "Monetary policy accounts"),
        (2, "BCE", "Discurso"),
    ],
    3: [
        (3, "BoE", "Decisión y actas MPC"),
        (3, "BoE", "Monetary Policy Report"),
        (3, "BoE", "Discurso"),
    ],
    4: [
        (3, "BoJ", "Comunicado de política monetaria"),
        (3, "BoJ", "Summary of Opinions"),
        (3, "BoJ", "Minutes"),
        (3, "BoJ", "Outlook Report"),
        (3, "BoJ", "Discurso original BoJ"),
        (5, "RBI (India)", "Comunicado y resolución MPC"),
        (5, "RBI (India)", "Monetary Policy Report"),
        (5, "PBoC (China)", "China Monetary Policy Report Q1 2025 (portal oficial regional)"),
    ],
    5: [
        (5, "BoC (Canadá)", "Decisión monetaria"),
        (5, "BoC (Canadá)", "Monetary Policy Report"),
        (5, "SNB (Suiza)", "Evaluación y decisión monetaria"),
        (5, "BCB (Brasil)", "Comunicado Copom"),
        (5, "Banxico (México)", "Anuncio de decisión"),
    ],
    6: [
        (6, "FMI", "World Economic Outlook: informe"),
        (10, "FMI", "GFSR: ficha de una edición"),
    ],
    7: [
        (8, "Fed", "Financial Stability Report"),
        (8, "BCE", "Financial Stability Review"),
        (8, "BoE", "Financial Stability Report"),
    ],
    8: [(9, "BIS", "Quarterly Review: edición completa")],
}

APARTADOS = {
    1: ("Fed", "Comunicación monetaria"),
    2: ("BCE", "Comunicación monetaria"),
    3: ("BoE", "Comunicación monetaria"),
    4: ("Asia: BoJ, PBoC y RBI", "Comunicación monetaria"),
    5: ("Canadá, Suiza, Brasil y México", "Comunicación monetaria"),
    6: ("FMI: WEO y GFSR", "Contexto macro y riesgo"),
    7: ("Estabilidad: Fed, BCE, BoE y BoJ", "Contexto de riesgo"),
    8: ("BIS Quarterly Review", "Contexto de mercados"),
}

def construir_inventario(base):
    por_clave = {(s["apartado"], s["fuente"], s["tipo"]): s for s in base}
    elegidas = []
    for apartado, claves in SELECCION.items():
        for clave in claves:
            if clave not in por_clave:
                raise ValueError(f"Revisar selección: falta la familia {clave}")
            spec = deepcopy(por_clave[clave])
            spec["apartado"] = apartado
            elegidas.append(spec)

    # Incorporaciones directas: proyecciones de los dos bancos principales y
    # estabilidad japonesa, para no limitar el riesgo sistémico a Occidente.
    elegidas.extend([
        {"apartado": 1, "fuente": "Fed", "tipo": "Summary of Economic Projections",
         "url": "https://www.federalreserve.gov/monetarypolicy/fomccalendars.htm",
         "patron": r"/monetarypolicy/fomcprojtabl\d{8}\.htm$",
         "adaptador": "documento", "aplicacion_relevante": False},
        {"apartado": 2, "fuente": "BCE", "tipo": "Proyecciones macroeconómicas",
         "url": "https://www.ecb.europa.eu/press/projections/html/index.en.html",
         "patron": r"/(?:pub|press)/projections/html/ecb\.projections\d{6}[^/]*\.en\.html$",
         "adaptador": "documento", "aplicacion_relevante": False},
        {"apartado": 7, "fuente": "BoJ", "tipo": "Financial System Report",
         "url": "https://www.boj.or.jp/en/research/brp/fsr/index.htm",
         "patron": [r"/research/brp/fsr/fsr\d{6}\.htm$",
                    r"/research/brp/fsr/data/fsr\d{6}a\.pdf$"],
         "adaptador": "documento", "aplicacion_relevante": False},
    ])

    # Sustituir muestras fijas por listados oficiales cuando se conoce su ruta.
    listados = {
        ("BCE", "Discurso"): (
            "https://www.ecb.europa.eu/press/key/html/index.en.html",
            r"/press/key/date/\d{4}/html/ecb\.sp\d{6}[^/]*\.en\.html$"),
        ("BoE", "Decisión y actas MPC"): (
            "https://www.bankofengland.co.uk/monetary-policy-summary-and-minutes/monetary-policy-summary-and-minutes",
            r"/monetary-policy-summary-and-minutes/\d{4}/[a-z]+-\d{4}$"),
        ("BoE", "Monetary Policy Report"): (
            "https://www.bankofengland.co.uk/sitemap/monetary-policy-report",
            r"/monetary-policy-report/\d{4}/[a-z]+-\d{4}$"),
        ("BoC (Canadá)", "Decisión monetaria"): (
            "https://www.bankofcanada.ca/core-functions/monetary-policy/key-interest-rate/",
            r"/\d{4}/\d{2}/fad-press-release-[^/]+/$"),
        ("BoC (Canadá)", "Monetary Policy Report"): (
            "https://www.bankofcanada.ca/publications/mpr/",
            r"/publications/mpr/mpr-\d{4}-\d{2}-\d{2}/$"),
        ("SNB (Suiza)", "Evaluación y decisión monetaria"): (
            "https://www.snb.ch/en/the-snb/mandates-goals/monetary-policy/decisions",
            r"/publications/communication/press-releases(?:-restricted)?/pre_\d{8}(?:_\d+)?$"),
    }
    for spec in elegidas:
        clave = spec["fuente"], spec["tipo"]
        if clave in listados:
            spec["url"], spec["patron"] = listados[clave]
        if clave == ("Fed", "Discurso"):
            spec["url"] = f"https://www.federalreserve.gov/newsevents/{pd.Timestamp(CORTE_UTC).year}-speeches.htm"
        spec["funcion"] = APARTADOS[spec["apartado"]][1]
        spec["alcance_acceso"] = (
            "Excluida de automatización" if spec.get("excluida") else
            "Listado: muestra limitada" if spec.get("patron") else
            "Edición fija: no acredita actualidad"
        )
    claves_finales = [(s["apartado"], s["fuente"], s["tipo"]) for s in elegidas]
    if len(claves_finales) != len(set(claves_finales)):
        raise ValueError("Hay familias duplicadas en la selección.")
    return sorted(elegidas, key=lambda s: s["apartado"])

sonda.inventory = construir_inventario(sonda.inventory)
inventario_activo = pd.DataFrame(sonda.inventory)
print(f"Selección activa: {len(inventario_activo)} familias; {len(APARTADOS)} apartados.")
print("Corte UTC:", CORTE_UTC)
display(inventario_activo[["apartado", "fuente", "tipo", "funcion", "alcance_acceso"]])
```

## Fechas, duplicación y lectura de resultados

**Fecha de reunión, publicación, aplicación y descarga son distintas.** En las actas se mide la publicación respecto a la reunión; la descarga de hoy no convierte un documento antiguo en información nueva. Una revisión de proyecciones se compara con su edición anterior, nunca con una versión revisada conocida después.

Se mantiene `sonda-timegate-v2`: texto completo, versión acreditada y disponibilidad anterior o igual al corte para la vista histórica. Una fecha diaria sin hora no permite simular una reacción intradía. Las traducciones pueden publicarse después del original. Una fecha ausente, un bloqueo o una extracción parcial quedan visibles.

La adquisición sigue consultando robots.txt y conserva las exclusiones existentes de PBoC y GFSR. No se reactivan descargas mediante rutas alternativas. **Bloqueada** significa falta de muestra, no ausencia de riesgo ni falta de importancia de la institución.

Cada apartado muestra inventario, cobertura, fechas, errores y gráficos obtenidos de su muestra. El cambio de vocabulario entre textos es descriptivo: no constituye sentimiento ni señal de compra o venta. La declaración y las preguntas de una misma comparecencia no son dos eventos independientes. Los originales se deduplican por hash al medir volumen global.

**Actualidad:** los adaptadores toman enlaces en el orden del listado y algunas familias conservan una edición fija. No hay garantía de obtener la última publicación. El panel final muestra la fecha más reciente **dentro de la muestra**, por familia, e identifica las muestras fijas y las ausencias. No debe presentarse como una clasificación automática de lo que domina el mercado hoy.

## 6. FMI: escenario macro y estabilidad global

<a id="fmi"></a>

## Panel de cobertura y antigüedad por publicación

La tabla incluye **todas las familias elegidas**, aunque no se hayan ejecutado o no se haya obtenido texto. La publicación más reciente se busca por familia, evitando que un discurso reciente oculte unas actas antiguas. Solo se admiten fechas de publicación con precisión diaria o de hora; se excluyen del panel las fechas posteriores al corte.

`dias_desde_publicacion` mide antigüedad de la muestra, no retraso de publicación ni ausencia de noticias más recientes. Una fecha reciente tampoco acredita que el documento fuese la última edición disponible. `texto_completo` indica si hay extracción completa de esa muestra; no acredita su versión histórica.

Para centrar el análisis: revisar primero decisiones y comparecencias nuevas; después el cambio respecto a la comunicación anterior de la misma familia; por último, contrastar los temas con expectativas de tipos, rendimientos, divisas y activos de la región. Este notebook no trae esos precios ni genera una conclusión automática sobre el régimen actual.

```python
def panel_actualidad(sonda, corte_utc):
    corte = pd.Timestamp(corte_utc)
    if corte.tzinfo is None:
        raise ValueError("El corte debe incluir zona horaria.")
    corte = corte.tz_convert("UTC")
    filas = []
    for spec in sonda.inventory:
        fila = {k: spec[k] for k in ["apartado", "fuente", "tipo", "funcion", "alcance_acceso"]}
        fila.update(estado="NO_EJECUTADA", publicacion_mas_reciente_muestra=pd.NaT,
                    dias_desde_publicacion=None, texto_completo=False, url=spec["url"])
        frame = sonda.results.get(spec["apartado"])
        if frame is not None:
            grupo = frame.loc[frame["fuente"].eq(spec["fuente"]) & frame["tipo"].eq(spec["tipo"])]
            if not grupo.empty:
                fila["estado"] = ", ".join(sorted(grupo["estado"].unique()))
                validas = grupo.loc[grupo["estado"].eq("OK") &
                                    grupo["precision_publicacion"].isin(["día", "hora"])].copy()
                validas["_publicacion"] = pd.to_datetime(
                    validas["fecha_publicacion"], utc=True, errors="coerce", format="mixed")
                validas = validas.loc[validas["_publicacion"].le(corte)]
                if not validas.empty:
                    ultima = validas.sort_values("_publicacion").iloc[-1]
                    fecha = ultima["_publicacion"]
                    fila.update(publicacion_mas_reciente_muestra=fecha,
                                dias_desde_publicacion=(corte - fecha).days,
                                texto_completo=bool(ultima["texto_completo"]), url=ultima["url"])
        if spec.get("excluida"):
            fila["estado"] = "EXCLUIDA_AUTOMATIZACION"
        filas.append(fila)
    return pd.DataFrame(filas)

actualidad = panel_actualidad(sonda, CORTE_UTC)
display(actualidad)
```
